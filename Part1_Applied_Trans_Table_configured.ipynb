## 📊 Applied_Trans Table Data

```python
"""
BigQuery -> GCS -> Local Parquet extraction utility.

Cost design:
- Uses BigQuery `extract_table` (an EXTRACT job), NOT a SQL query. Extract jobs
  are free of BigQuery query/slot charges - you are only billed for GCS storage
  (briefly, since we delete it right after) and network egress when downloading
  the file(s) to your machine.
- Dataset location is auto-detected so the extract job and GCS bucket are kept
  in the same region, avoiding cross-region data-movement charges.
- Large tables (yours is ~3.19 GB) cannot be extracted to a single file - BigQuery
  caps single-file extracts at ~1 GB compressed. So we extract with a wildcard
  URI (sharded), download all shards, concatenate into one parquet file locally,
  then delete the GCS objects and local shard files.

NEW in this version:
- `list_local_parquet_files()` - see every parquet file currently sitting in
  your local ./data folder.
- `select_parquet_file()` - pick out exactly one of those files by matching a
  fragment of its filename (e.g. a date suffix), so if you have both
  `..._20260707_trans.parquet` and `..._20260708_trans.parquet` you can choose
  precisely which one gets loaded into BigQuery via create_table_from_parquet().

Install requirements (if not already available):
    pip install google-cloud-bigquery google-cloud-storage pyarrow
"""

import os
import glob
from datetime import datetime, timedelta

from google.cloud import bigquery, storage
import pyarrow.parquet as pq
import pyarrow as pa


def data_extraction(
    table_name: str,
    output_file_prefix: str,
    project: str = "prj-prod-dataplatform",
    bucket: str = "prod-asia-southeast1-tonik-aiml-workspace",
    gcs_folder: str = "DC/Parquet_files",
    date_suffix: str = None,
    keep_gcs_copy: bool = False,
):
    """
    Extracts a BigQuery table to Parquet via GCS (least-cost path), downloads
    it to <cwd>/data/<output_file_prefix><date_suffix>.parquet, and cleans up
    the GCS copy afterward.

    Parameters
    ----------
    table_name : str
        'dataset.table', e.g. 'risk_mart.applied_loans_trans'
    output_file_prefix : str
        e.g. 'risk_mart.applied_loans_20230101_'  (the date suffix is appended)
    project : str
        GCP project id running the BigQuery job / owning the bucket.
    bucket : str
        GCS bucket name (no 'gs://' prefix).
    gcs_folder : str
        Folder path inside the bucket to stage the temp parquet shards.
    date_suffix : str, optional
        Defaults to yesterday's date (today - 1) as YYYYMMDD, per your convention.
    keep_gcs_copy : bool
        If True, skips deleting the GCS files after download (default False).

    Returns
    -------
    str : local path to the final combined parquet file.
    """

    # ---- 1. Resolve names/paths ----
    if date_suffix is None:
        date_suffix = (datetime.today() - timedelta(days=1)).strftime("%Y%m%d")

    output_filename = f"{output_file_prefix}{date_suffix}_trans.parquet"

    local_dir = os.path.join(os.getcwd(), "data")
    os.makedirs(local_dir, exist_ok=True)
    local_path = os.path.join(local_dir, output_filename)

    dataset_id, table_id = table_name.split(".")

    # ---- 2. Set up clients ----
    # bq_client = bigquery.Client(project=project)
    bq_client = client
    gcs_client = storage.Client(project=project)

    # Auto-detect dataset location to avoid cross-region egress charges
    dataset_ref = bigquery.DatasetReference(project, dataset_id)
    dataset = bq_client.get_dataset(dataset_ref)
    location = dataset.location
    print(f"Dataset location detected: {location}")

    table_ref = bigquery.TableReference(dataset_ref, table_id)

    # ---- 3. Extract table -> GCS (wildcard, since table > ~1GB) ----
    gcs_object_prefix = f"{gcs_folder}/{output_file_prefix}{date_suffix}"
    destination_uri = f"gs://{bucket}/{gcs_object_prefix}_*.parquet"

    extract_config = bigquery.ExtractJobConfig(
        destination_format=bigquery.DestinationFormat.PARQUET
    )

    print(f"Starting extract job: {table_name} -> {destination_uri}")
    extract_job = bq_client.extract_table(
        table_ref,
        destination_uri,
        location=location,
        job_config=extract_config,
    )
    extract_job.result()  # wait for completion
    print("Extract job complete.")

    # ---- 4. Download shard files from GCS ----
    blobs = list(gcs_client.list_blobs(bucket, prefix=gcs_object_prefix))
    if not blobs:
        raise RuntimeError(
            f"No files found at gs://{bucket}/{gcs_object_prefix}_*.parquet after extract job."
        )

    print(f"Found {len(blobs)} shard file(s) in GCS. Downloading...")
    local_shard_paths = []
    for blob in blobs:
        shard_local_path = os.path.join(local_dir, os.path.basename(blob.name))
        blob.download_to_filename(shard_local_path)
        local_shard_paths.append(shard_local_path)
    print("Download complete.")

    # ---- 5. Combine shards into a single parquet file ----
    if len(local_shard_paths) == 1:
        os.replace(local_shard_paths[0], local_path)
    else:
        print(f"Combining {len(local_shard_paths)} shards into {output_filename}...")
        tables = [pq.read_table(p) for p in local_shard_paths]
        combined = pa.concat_tables(tables)
        pq.write_table(combined, local_path)
        # remove the individual shard files, keep only the combined file
        for p in local_shard_paths:
            os.remove(p)

    print(f"Final file saved locally at: {local_path}")

    # ---- 6. Delete GCS copies to avoid storage cost ----
    if not keep_gcs_copy:
        print("Deleting temporary GCS files to save storage cost...")
        for blob in blobs:
            blob.delete()
        print("GCS cleanup complete.")
    else:
        print("keep_gcs_copy=True -> leaving files in GCS.")

    return local_path


def list_local_parquet_files(local_dir: str = None, name_contains: str = None):
    """
    Lists parquet files currently sitting in your local folder, so you can see
    what's available before picking one to load into BigQuery.

    Parameters
    ----------
    local_dir : str, optional
        Defaults to <cwd>/data (same folder data_extraction() writes to).
    name_contains : str, optional
        If given, only filenames containing this substring are returned
        (e.g. 'applied_loans_20230101' to narrow down to one table family).

    Returns
    -------
    list[str] : sorted list of matching local file paths.
    """
    if local_dir is None:
        local_dir = os.path.join(os.getcwd(), "data")

    all_files = sorted(glob.glob(os.path.join(local_dir, "*.parquet")))
    if name_contains:
        all_files = [f for f in all_files if name_contains in os.path.basename(f)]

    return all_files


def select_parquet_file(match: str, local_dir: str = None):
    """
    Selects exactly one local parquet file whose filename contains `match`.

    This is the piece that lets you disambiguate when you have several dated
    extracts sitting in the same folder, e.g.:
        risk_mart.applied_loans_20230101_20260707_trans.parquet
        risk_mart.applied_loans_20230101_20260708_trans.parquet
    Calling select_parquet_file("20260707") returns the path to the first one
    only. Pass a longer/more specific fragment (e.g. "20230101_20260707_trans")
    if a short fragment could match more than one file.

    Parameters
    ----------
    match : str
        Substring to search for in the filename (not a glob pattern).
    local_dir : str, optional
        Defaults to <cwd>/data.

    Returns
    -------
    str : the single matching local file path.

    Raises
    ------
    FileNotFoundError
        If no file matches `match`.
    ValueError
        If more than one file matches `match` (so you never silently load
        the wrong table).
    """
    if local_dir is None:
        local_dir = os.path.join(os.getcwd(), "data")

    candidates = [
        f for f in sorted(glob.glob(os.path.join(local_dir, "*.parquet")))
        if match in os.path.basename(f)
    ]

    if not candidates:
        available = list_local_parquet_files(local_dir)
        raise FileNotFoundError(
            f"No parquet file matching '{match}' found in {local_dir}.\n"
            f"Available files:\n" + "\n".join(available)
        )

    if len(candidates) > 1:
        raise ValueError(
            f"Multiple parquet files match '{match}', please use a more "
            f"specific fragment:\n" + "\n".join(candidates)
        )

    return candidates[0]


def create_table_from_parquet(
    local_file_path: str,
    table_name: str = None,
    project: str = "prj-prod-dataplatform",
    location: str = None,
    write_disposition: str = "WRITE_TRUNCATE",
):
    """
    Loads a local parquet file into BigQuery as a new table, using the schema
    embedded in the parquet file (no manual schema definition needed).

    This uses a LOAD job, which - like the extract job - is free of BigQuery
    query/slot charges. You only pay for the storage the resulting table uses,
    which is fine here since you'll drop the table when you're done with it.

    Parameters
    ----------
    local_file_path : str
        Path to the local .parquet file, e.g.
        './data/risk_mart.applied_loans_20230101_20260707_trans.parquet'
        Use select_parquet_file() to pick this out safely when your ./data
        folder has more than one candidate file.
    table_name : str, optional
        'dataset.table' to create. If not provided, it's derived from the
        filename itself (stripping the .parquet extension), matching your
        convention of naming the table the same as the file.
    project : str
        GCP project id that will own the table.
    location : str, optional
        BigQuery region for the destination dataset. If not given, it's
        auto-detected from the existing dataset (must already exist).
    write_disposition : str
        'WRITE_TRUNCATE' (default, overwrite if table exists),
        'WRITE_EMPTY' (fail if table exists), or 'WRITE_APPEND'.

    Returns
    -------
    str : fully-qualified table id created, e.g.
          'prj-prod-dataplatform.risk_mart.applied_loans_20230101_20260707_trans'
    """

    if table_name is None:
        base_name = os.path.splitext(os.path.basename(local_file_path))[0]
        table_name = base_name  # expects 'dataset.table' pattern already in filename

    dataset_id, table_id = table_name.split(".", 1)

    bq_client = bigquery.Client(project=project)

    dataset_ref = bigquery.DatasetReference(project, dataset_id)
    if location is None:
        dataset = bq_client.get_dataset(dataset_ref)  # dataset must already exist
        location = dataset.location
    print(f"Target dataset location: {location}")

    table_ref = bigquery.TableReference(dataset_ref, table_id)
    full_table_id = f"{project}.{dataset_id}.{table_id}"

    job_config = bigquery.LoadJobConfig(
        source_format=bigquery.SourceFormat.PARQUET,
        write_disposition=write_disposition,
    )

    print(f"Loading {local_file_path} -> {full_table_id} ...")
    with open(local_file_path, "rb") as source_file:
        load_job = bq_client.load_table_from_file(
            source_file, table_ref, job_config=job_config, location=location
        )
    load_job.result()  # wait for completion

    table = bq_client.get_table(table_ref)
    print(f"Table created: {full_table_id}")
    print(f"Rows loaded: {table.num_rows}")
    print(f"Schema fields: {len(table.schema)}")

    return full_table_id


def drop_table(table_name: str, project: str = "prj-prod-dataplatform"):
    """
    Convenience helper to delete the temp table once you're done with it.
    table_name : 'dataset.table'
    """
    bq_client = bigquery.Client(project=project)
    full_table_id = f"{project}.{table_name}"
    bq_client.delete_table(full_table_id, not_found_ok=True)
    print(f"Deleted table: {full_table_id}")






    # When you're done with the temp table:
    # drop_table("risk_mart.applied_loans_20230101_20260707_trans")
```
