```python
def save_to_gcs(data, filename, bucket_name):
    """
    Save data to Google Cloud Storage bucket.

    :param data: The data to save. Can be a string, bytes, or a file-like object.
    :param filename: The name of the file to save in the bucket.
    :param bucket_name: The name of the GCS bucket. Defaults to 'ABC'.
    :return: The public URL of the saved file.
    """
    # Create a client
    client = storage.Client()

    # Get the bucket
    bucket = client.bucket(bucket_name)

    # Create a blob (file) in the bucket
    blob = bucket.blob(filename)

    # Determine the content type and upload accordingly
    if isinstance(data, str):
        blob.upload_from_string(data)
    elif isinstance(data, bytes):
        blob.upload_from_string(data, content_type='application/octet-stream')
    elif hasattr(data, 'read'):  # File-like object
        blob.upload_from_file(data)
    else:
        raise ValueError("Unsupported data type. Please provide a string, bytes, or file-like object.")


    print(f"File {filename} uploaded to {bucket_name}.")
    return blob.public_url



def load_pickled_dict_from_gcs(filename, bucket_name, project_id):

    client = storage.Client(project=project_id)

    bucket = client.bucket(bucket_name)
    blob = bucket.blob(filename)

    data = blob.download_as_bytes()

    import pickle
    return pickle.loads(data)



def load_csv_from_gcs(filename, bucket_name):

    return pd.read_csv(
        generate_bucket_url(filename, bucket_name)
    )


def generate_bucket_url(filename, bucket_name):

    return f"gs://{bucket_name}/{filename}"


def union_dict_values(input_dict: dict) -> list:
    """
    Create a unique list of all values from a dictionary of lists.

    Args:
        input_dict (dict): A dictionary where values are lists.

    Returns:
        list: A sorted list of unique values from all lists in the input dictionary.
    """
    # Use a set comprehension to flatten and deduplicate values
    unique_values = {item for sublist in input_dict.values() for item in sublist}

    # Convert the set to a sorted list
    return sorted(unique_values)


def add_prefix_to_columns(df, prefix, columns=None):
    """
    Add a prefix to column names in a pandas DataFrame.

    Parameters:
    df (pandas.DataFrame): The input DataFrame
    prefix (str): The prefix to add to the column names
    columns (list, optional): List of column names to add the prefix to.
                              If None, prefix is added to all columns.

    Returns:
    pandas.DataFrame: A new DataFrame with updated column names
    """
    # If columns is None, use all columns
    if columns is None:
        columns = df.columns

    # Create a dictionary mapping old column names to new prefixed names
    rename_dict = {col: f"{prefix}{col}" for col in columns if col in df.columns}

    # Return a new DataFrame with renamed columns
    return df.rename(columns=rename_dict, inplace=True)


def merge_category_dataframes(base_df, cust_cat_dict):
    """
    Merge dataframes from cust_cat_dict into the base dataset as separate columns.

    Parameters:
    base_df (pandas.DataFrame): The base dataset with 'package_name' column
    cust_cat_dict (dict): Dictionary with tag names as keys and dataframes as values

    Returns:
    pandas.DataFrame: A new DataFrame with additional columns from cust_cat_dict
    """
    if 'package_name' not in base_df.columns:
        raise KeyError("Base DataFrame is missing 'package_name' column")

    result_df = base_df.copy()

    for tag, df in cust_cat_dict.items():
        if 'package_name' not in df.columns:
            raise KeyError(f"DataFrame for tag '{tag}' is missing 'package_name' column")

        if len(df.columns) != 1:
            raise ValueError(f"DataFrame for tag '{tag}' should have only one column (package_name)")

        # Create a new column with the tag name, containing 1 for packages present
        df_merged = df.assign(**{tag: 1})

        # Merge while keeping the 'package_name' column intact
        result_df = result_df.merge(df_merged, on='package_name', how='left')

        # Fill NaN values with 0 and convert to int
        result_df[tag] = result_df[tag].fillna(0).astype(int)

    return result_df


def merge_category_dataframes_optimized(base_df, cust_cat_dict):
    """
    Merge dataframes from cust_cat_dict into the base dataset as separate columns,
    handling potential duplicates without creating a copy of the base DataFrame.

    Parameters:
    base_df (pandas.DataFrame): The base dataset with 'package_name' column
    cust_cat_dict (dict): Dictionary with tag names as keys and dataframes as values

    Returns:
    pandas.DataFrame: The original DataFrame with additional columns from cust_cat_dict
    """
    if 'package_name' not in base_df.columns:
        raise KeyError("Base DataFrame is missing 'package_name' column")

    for tag, df in cust_cat_dict.items():
        if 'package_name' not in df.columns:
            raise KeyError(f"DataFrame for tag '{tag}' is missing 'package_name' column")

        if len(df.columns) != 1:
            raise ValueError(f"DataFrame for tag '{tag}' should have only one column (package_name)")

        # Create a new Series with 1 for packages present, handling duplicates
        category_series = pd.Series(1, index=df['package_name']).groupby(level=0).first()

        # Use map with a default of 0, and convert to int8
        base_df[tag] = base_df['package_name'].map(category_series).fillna(0).astype('int8')

    return base_df
```

## Metadata functions

```python
def add_metadata_columns(df, metadata_dict, metadata_keys, col_prefix):
    """
    Add metadata columns to a DataFrame based on package names, modifying the DataFrame in place.

    Args:
        df (pd.DataFrame): DataFrame with a 'package_name' column.
        metadata_dict (dict): Dictionary of package metadata.
        metadata_keys (list): List of metadata keys to add as columns.
        col_prefix (str): Prefix for new columns

    Warns:
        UserWarning: If any metadata keys are not found.
    """
    # Preserve string-capable dtypes when every loaded app is ineligible.
    if df.empty:
        for key in metadata_keys:
            df[f'{col_prefix}{key}'] = pd.Series(index=df.index, dtype='object')
        return

    # Function to get metadata for a package, returning None if not found
    def get_metadata(package, key):
        return metadata_dict.get(package, {}).get(key)

    # Set to keep track of keys not found in any package
    missing_keys = set(metadata_keys)

    # Calculate total iterations for the progress bar
    total_iterations = len(metadata_keys) * len(df)

    # Create a single progress bar for the entire process
    with tqdm(total=total_iterations, desc="Processing metadata") as pbar:
        # Add new columns for each metadata key
        for key in metadata_keys:
            key_found = False
            new_column = []
            for package in df['package_name']:
                value = get_metadata(package, key)
                new_column.append(value)
                if value is not None:
                    key_found = True
                pbar.update(1)  # Update progress bar for each package

            df[f'{col_prefix}{key}'] = new_column
            if key_found:
                missing_keys.discard(key)

    # Warn about keys not found in any package
    if missing_keys:
        warnings.warn(f"The following keys were not found in any package's metadata: {', '.join(missing_keys)}")


def extract_first_category_id(df, column_name):
    """
    Extract the ID of the first category from a column containing a list of category dictionaries.

    Args:
        df (pd.DataFrame): Input DataFrame.
        column_name (str): Name of the column containing category information.

    Returns:
        pd.Series: Series with the first category ID or Unknown (in case of category ID = None) or None values.
    """
    error_count = 0

    def get_first_id(value):
        nonlocal error_count
        try:
            if value:
                result = value[0]['id'].lower()
                if result:
                    return result
                else:
                    return "Unknown"
            else:
                return None
        except:
            # Increment error count and return NA if there's any error in parsing or accessing the data
            error_count += 1
            return None

    result = df[column_name].apply(get_first_id)

    print(f"Total errors encountered during parsing or accessing data: {error_count}")

    return result

def extract_first_category_name(df, column_name):
    """
    Extract the name of the first category from a column containing a list of category dictionaries.

    Args:
        df (pd.DataFrame): Input DataFrame.
        column_name (str): Name of the column containing category information.

    Returns:
        pd.Series: Series with the first category name or Unknown (in case of category name = None) or None values.
    """
    error_count = 0

    def get_first_name(value):
        nonlocal error_count
        try:
            if value:
                result = value[0]['name'].lower()
                if result:
                    return result
                else:
                    return "Unknown"
            else:
                return None
        except:
            # Increment error count and return NA if there's any error in parsing or accessing the data
            error_count += 1
            return None

    result = df[column_name].apply(get_first_name)

    print(f"Total errors encountered during parsing or accessing data: {error_count}")

    return result
```

```python
import re

def to_snake_case(value):
    """Convert category name to lowercase snake_case."""
    value = str(value).strip()

    # AbcXyz -> Abc_Xyz
    value = re.sub(r"(?<=[a-z0-9])(?=[A-Z])", "_", value)

    # Spaces, hyphens and other separators -> _
    value = re.sub(r"[^a-zA-Z0-9]+", "_", value)

    # Remove duplicate/edge underscores
    value = re.sub(r"_+", "_", value).strip("_")

    return value.lower()


def add_custom_category_flags(
    df,
    category_df,
    prefix,
    package_col="package_name",
    category_col="ps_custom_category",
):
    """
    Add one-hot encoded category flags to the main DataFrame based on package name.

    Parameters
    ----------
    df : pd.DataFrame
        Main DataFrame. `package_col` does not have to be unique.

    category_df : pd.DataFrame
        Lookup DataFrame with unique package names and their category.

    prefix : str
        Prefix added to generated category flag columns.
        Example: prefix="ps" -> ps_finance, ps_gambling, ...

    package_col : str, default="package_name"
        Package name column used for joining.

    category_col : str, default="ps_custom_category"
        Column containing category labels.

    Returns
    -------
    pd.DataFrame
        Original DataFrame with additional binary category flag columns.
    """

    # Validate uniqueness of the lookup table
    if category_df[package_col].duplicated().any():
        raise ValueError(
            f"`{package_col}` must be unique in category_df."
        )

    # Keep only required columns
    lookup = category_df[[package_col, category_col]].copy()
    lookup[category_col] = lookup[category_col].map(to_snake_case)

    # Create one-hot encoded category columns
    category_flags = pd.get_dummies(
        lookup[category_col],
        prefix=prefix,
        prefix_sep="_",
        dtype=int,
    )

    # Attach package name back to the encoded table
    lookup_flags = pd.concat(
        [
            lookup[[package_col]].reset_index(drop=True),
            category_flags.reset_index(drop=True),
        ],
        axis=1,
    )

    # Merge flags into the main dataset
    result = df.merge(
        lookup_flags,
        on=package_col,
        how="left",
        validate="many_to_one",
    )

    # Packages not found in the lookup should have 0 for all category flags
    flag_cols = category_flags.columns.tolist()

    result[flag_cols] = (
        result[flag_cols]
        .fillna(0)
        .astype(int)
    )

    return result


def rename_categories(df, column, mapping_dict, default_category):
    df[f'{column}_mapped'] = df[column].map(mapping_dict).fillna(default_category).copy()
    return df

def aggregate_flag_groups(df, grouping_dict):
    for group_name, columns in grouping_dict.items():
        missing_columns = [col for col in columns if col not in df.columns]

        if missing_columns:
            print(f"{group_name}: missing columns -> {missing_columns}")

        existing_columns = [col for col in columns if col in df.columns]

        if existing_columns:
            row_sum = df[existing_columns].sum(axis=1)
            df[group_name] = (row_sum > 0).astype(int)
        else:
            df[group_name] = 0

    return df
```

```python
def check_app_status_vectorized(flags):
    """
    Vectorized function to determine if applications are system (1) or non-system (0)
    based on their flag values. Handles NA values by returning np.nan.

    Args:
        flags (pd.Series or np.array or list or int): Application flag values

    Returns:
        pd.Series or int: 1 (system) for odd numbers, 0 (non-system) for even numbers,
                         np.nan for NA/null values
    """
    if isinstance(flags, (pd.Series, np.ndarray, list)):
        # Convert to pandas Series if not already
        flags = pd.Series(flags)

        # Initialize result series with NaN values
        result = pd.Series([np.nan] * len(flags), index=flags.index)

        # Update non-null values
        mask = flags.notna()
        result[mask] = np.where(flags[mask].astype(float) % 2 == 1, 1, 0)

        return result
    else:
        # Handle single value
        if pd.isna(flags):
            return np.nan
        return 1 if int(flags) % 2 == 1 else 0
```

```python
valid_installs = app_feature_rows  # Already deduplicated and as-of filtered.
install_dynamics_dt = inventory_coverage_dt[[loan_id]].set_index(loan_id)
valid_count = inventory_coverage_dt.set_index(loan_id)['install_ts_valid_cnt'].replace(0, np.nan)
age_groups = valid_installs.groupby(loan_id)['install_age_days']

for days in (3, 7, 30, 90, 180, 365):
    recent_count = (
        valid_installs['install_age_days'].lt(days)
        .groupby(valid_installs[loan_id]).sum()
    )
    install_dynamics_dt[f'installs_cnt_{days}d'] = recent_count.reindex(
        install_dynamics_dt.index, fill_value=0
    ).astype(int)
    if days != 1:
        install_dynamics_dt[f'installs_share_{days}d'] = (
            install_dynamics_dt[f'installs_cnt_{days}d'] / valid_count
        )

for percentile in (10, 25, 50, 75, 90):
    install_dynamics_dt[f'user_install_age_q{percentile}_days'] = (
        age_groups.quantile(percentile / 100)
    )
install_dynamics_dt['user_install_age_iqr_days'] = (
    install_dynamics_dt['user_install_age_q75_days']
    - install_dynamics_dt['user_install_age_q25_days']
)
install_dynamics_dt['user_install_age_gt365_share'] = (
    valid_installs['install_age_days'].gt(365)
    .groupby(valid_installs[loan_id]).sum()
    .reindex(install_dynamics_dt.index, fill_value=0) / valid_count
)

for days in (30, 90):
    recent = valid_installs.loc[valid_installs['install_age_days'].lt(days)]
    install_dynamics_dt[f'active_install_days_{days}d'] = (
        recent.groupby(loan_id)['install_day'].nunique()
        .reindex(install_dynamics_dt.index, fill_value=0).astype(int)
    )
    install_dynamics_dt[f'max_installs_same_day_{days}d'] = (
        recent.groupby([loan_id, 'install_day']).size().groupby(level=0).max()
        .reindex(install_dynamics_dt.index, fill_value=0).astype(int)
    )

recent_year = valid_installs.loc[valid_installs['install_age_days'].lt(365)].copy()
recent_year['install_month'] = recent_year['install_day'].dt.to_period('M')
install_dynamics_dt['active_install_months_365d'] = (
    recent_year.groupby(loan_id)['install_month'].nunique()
    .reindex(install_dynamics_dt.index, fill_value=0).astype(int)
)

daily_installs = (
    valid_installs.groupby([loan_id, 'install_day']).size()
    .rename('daily_count').reset_index()
)
largest_day = (
    daily_installs.sort_values(
        [loan_id, 'daily_count', 'install_day'], ascending=[True, False, False]
    )
    .drop_duplicates(loan_id).set_index(loan_id)
)
asof_by_loan = app_feature_rows.drop_duplicates(loan_id).set_index(loan_id)['asof_day']
install_dynamics_dt['largest_burst_recency_days'] = (
    asof_by_loan.reindex(largest_day.index) - largest_day['install_day']
).dt.days


prior_8to30 = install_dynamics_dt['installs_cnt_30d'] - install_dynamics_dt['installs_cnt_7d']
prior_31to180 = install_dynamics_dt['installs_cnt_180d'] - install_dynamics_dt['installs_cnt_30d']
install_dynamics_dt['install_velocity_7_vs_30'] = (
    (install_dynamics_dt['installs_cnt_7d'] / 7)
    / (prior_8to30 / 23).replace(0, np.nan)
)
install_dynamics_dt['install_velocity_30_vs_180'] = (
    (install_dynamics_dt['installs_cnt_30d'] / 30)
    / (prior_31to180 / 150).replace(0, np.nan)
)
install_dynamics_dt['install_prior_8to30_zero_flag'] = prior_8to30.eq(0).astype(int)
install_dynamics_dt['install_prior_31to180_zero_flag'] = prior_31to180.eq(0).astype(int)
for tau in (30, 90):
    weight = np.exp(-valid_installs['install_age_days'] / tau)
    install_dynamics_dt[f'recency_weighted_installs_tau{tau}'] = (
        weight.groupby(valid_installs[loan_id]).sum()
        .reindex(install_dynamics_dt.index, fill_value=0)
    )
install_dynamics_dt = install_dynamics_dt.reset_index()
```

```python
def create_mapping_dict(columns, pattern='app_cnt_{name}_ever'):
    """
    Replace '*_cat_' or 'cat_' prefix with pattern in column names.

    Parameters:
    columns (list): List of column names
    pattern (str): Pattern to use for new names, must contain {name} placeholder
        default: 'app_cnt_{name}_ever'

    Returns:
    dict: Dictionary mapping old column names to new ones
    """
    rename_dict = {}

    for col in columns:
        if '_cat_' in col:
            new_name = col.split('_cat_')[1]

            # EDIT 2026-10-06
            prefix = col.split('_cat_')[0]
            if prefix == "llm":
                new_name = f"llm_{new_name}"

            new_name = pattern.format(name=new_name)
            rename_dict[col] = new_name
        elif col.startswith('cat_'):
            new_name = col.replace('cat_', '', 1)  # just remove 'cat_'
            new_name = pattern.format(name=new_name)
            rename_dict[col] = new_name
        else:
            rename_dict[col] = col

    return rename_dict

def get_first_values_by_loan(df, category_columns, id_column):
    """
    Get first values from multiple columns for each loan.

    Parameters:
    df (pandas.DataFrame): DataFrame containing:
        - id_column
        - columns (columns to get first values from)
    columns (list): List of column names to get first values

    Returns:
    pandas.DataFrame: DataFrame with first values from specified columns for each loan
    """
    return df.groupby(id_column)[category_columns].first().reset_index()
```

```python
def get_packages_by_loan(df, package_col, id_column):
    """
    Get list of packages for each loan.

    Parameters:
    df (pandas.DataFrame): DataFrame containing:
        - id_column
        - package_col (column with packages)
    columns (list): List of column names to get first values

    Returns:
    pandas.DataFrame: DataFrame with list of packages for each loan
    """
    return df.groupby(id_column)[package_col].agg(list).reset_index()

# new
def get_package_install_times_by_loan(df, package_col, time_col, id_column):
    """
    Faster version: aggregate to lists, then build dicts once per loan.
    """
    # 1) Aggregate packages and times into lists per loan
    agg = (
        df
        .groupby(id_column, sort=False)[[package_col, time_col]]
        .agg(list)
    )

    # 2) Build {package: time} dict for each loan
    agg['package_install_times'] = [
        dict(zip(pkgs, times))
        for pkgs, times in zip(agg[package_col], agg[time_col])
    ]

    # 3) Return clean result
    return (
        agg
        .drop(columns=[package_col, time_col])
        .reset_index()
    )
```

```python
def aggregate_categories_single_column(df, col_with_categories, id_column):
    """
    Aggregate package counts by category (single column) at the loan level.

    Parameters:
    df (pandas.DataFrame): DataFrame containing:
        - id_column
        - package_name
        - col_with_categories

    Returns:
    pandas.DataFrame: Aggregated DataFrame with:
        - id_column as index
        - Columns for each general category containing package counts
    """
    # First, get unique combinations of loan and category with their counts
    category_counts = (df.groupby([id_column, col_with_categories])
                      .package_name
                      .count()
                      .reset_index())

    # Pivot the results to get categories as columns
    result = category_counts.pivot(
        index=id_column,
        columns=col_with_categories,
        values='package_name'
    ).fillna(0)

    # Convert counts to integers
    result = result.astype(int)

    # Reset index to make id_column a column
    result = result.reset_index()

    # Rename columns to avoid potential naming conflicts
    category_columns = {col: f'app_cnt_{col}_ever' for col in result.columns
                       if col != id_column}
    result = result.rename(columns=category_columns)

    return result.reset_index(drop=True)

def aggregate_categories_different_columns(df, category_columns, id_column):
    """
    Aggregate package counts by categories (multiple columns) at the loan level.

    Parameters:
    df (pandas.DataFrame): DataFrame containing:
        - id_column
        - category columns (binary 0/1 values)
    category_columns (list): List of category column names to aggregate

    Returns:
    pandas.DataFrame: Aggregated DataFrame with counts for each category
    """
    return df.groupby(id_column)[category_columns].sum().reset_index()
```

## Category shares, recency and distinct-package interactions


```python
loan_index = inventory_coverage_dt.set_index(loan_id).index

all_apps_denominator = (
    inventory_coverage_dt.set_index(loan_id)['install_ts_valid_cnt']
    .replace(0, np.nan)
)
recent_apps_denominator = (
    install_dynamics_dt.set_index(loan_id)['installs_cnt_30d']
    .reindex(loan_index)
    .replace(0, np.nan)
)

custom_counts_ever = custom_categories_dt.set_index(loan_id)
custom_counts_30d = custom_categories_30d_dt.set_index(loan_id)

ever_cols = [
    col for col in custom_counts_ever.columns
    if col.startswith('app_cnt_') and col.endswith('_ever')
]
recent_cols = [
    col for col in custom_counts_30d.columns
    if col.startswith('app_cnt_') and col.endswith('_30d')
]
recent_names = [
    col[len('app_cnt_'):-len('_30d')] for col in recent_cols
]

payday_names = ('payday', 'llm_payday')
gambling_names = ('gambling', 'llm_gambling', 'llm_gambling_brands')

# Every category used below needs a package-level source flag.
required_names = set(recent_names) | set(payday_names) | set(gambling_names)
missing_names = required_names - custom_profile_tags.keys()
if missing_names:
    raise KeyError(f'Missing categories in custom_profile_tags: {sorted(missing_names)}')

missing_cols = {
    custom_profile_tags[name] for name in required_names
} - set(app_feature_rows.columns)
if missing_cols:
    raise KeyError(f'Missing package-level columns in app_feature_rows: {sorted(missing_cols)}')

# Calculate all shares as DataFrames instead of inserting columns one by one.
ever_shares = (
    custom_counts_ever[ever_cols]
    .reindex(loan_index)
    .fillna(0)
    .div(all_apps_denominator, axis=0)
)
ever_shares.columns = [
    col.replace('app_cnt_', 'app_share_', 1) for col in ever_cols
]

recent_shares = (
    custom_counts_30d[recent_cols]
    .reindex(loan_index)
    .fillna(0)
    .div(recent_apps_denominator, axis=0)
)
recent_shares.columns = [
    col.replace('app_cnt_', 'app_share_', 1) for col in recent_cols
]

# Collect recency Series first, then create one DataFrame.
recency_features = {}
for name in recent_names:
    tag_col = custom_profile_tags[name]
    tagged = pd.to_numeric(
        app_feature_rows[tag_col], errors='coerce'
    ).fillna(0).gt(0)

    recency_features[f'app_{name}_recency_days'] = (
        app_feature_rows.loc[tagged]
        .groupby(loan_id)['install_age_days']
        .min()
    )

recency_dt = pd.DataFrame(recency_features, index=loan_index)

# OR within each tag family; AND between the two families.
payday_cols = [custom_profile_tags[name] for name in payday_names]
gambling_cols = [custom_profile_tags[name] for name in gambling_names]

tag_values = (
    app_feature_rows[payday_cols + gambling_cols]
    .apply(pd.to_numeric, errors='coerce')
    .fillna(0)
    .gt(0)
)
payday_mask = tag_values[payday_cols].any(axis=1)
gambling_mask = tag_values[gambling_cols].any(axis=1)
relevant_mask = payday_mask | gambling_mask

tagged_apps = app_feature_rows.loc[
    relevant_mask, [loan_id, 'package_name', 'install_age_days']
].copy()
tagged_apps['payday'] = payday_mask.loc[relevant_mask].to_numpy()
tagged_apps['gambling'] = gambling_mask.loc[relevant_mask].to_numpy()


def distinct_payday_gambling_flag(rows):
    """Require payday and gambling tags on two distinct package names."""
    by_loan = rows.groupby(loan_id).agg(
        has_payday=('payday', 'any'),
        has_gambling=('gambling', 'any'),
        tagged_package_cnt=('package_name', 'nunique'),
    )
    return (
        by_loan['has_payday']
        & by_loan['has_gambling']
        & by_loan['tagged_package_cnt'].ge(2)
    ).reindex(loan_index, fill_value=False).astype('int8')


recent_tagged_apps = tagged_apps.loc[tagged_apps['install_age_days'].lt(30)]

interaction_dt = pd.DataFrame({
    'app_payday_x_gambling_presence':
        distinct_payday_gambling_flag(tagged_apps),
    'app_recent_payday_x_gambling_30d':
        distinct_payday_gambling_flag(recent_tagged_apps),
}, index=loan_index)

custom_profile_dt = pd.concat(
    [ever_shares, recent_shares, recency_dt, interaction_dt],
    axis=1,
).reset_index()
```

## First-install ages and installation gaps


```python
def calculate_time_since_first_app(df, category_column, id_column,
                                   result_column_name='days_since_first_app'):
    """Calendar days between inquiry and the first app in a category."""
    # EDIT 2026-09-30: The oldest app has the largest shared install age.
    age = (
        df.loc[df[category_column] == 1]
        .groupby(id_column)['app_install_age_days'].max()
    )
    return age.rename(result_column_name).reset_index()
```

```python
def calculate_installation_intervals_alltime(df, id_column):
    """
    Calculate average and median time between app installations for all time,
    excluding the first (oldest) installation time.

    Parameters:
    df (pandas.DataFrame): DataFrame containing:
        - id_column
        - first_install_time

    Returns:
    pandas.DataFrame: DataFrame with intervals calculations
    """
    # Sort only necessary columns to reduce memory footprint
    sorted_df = (df[[id_column, 'first_install_time']]
                 .sort_values([id_column, 'first_install_time']))

    # Calculate time differences and convert to minutes in one step
    intervals_mins = (sorted_df
                      .groupby(id_column, sort=False)['first_install_time']
                      .diff()
                      .dt.total_seconds()
                      .div(60))

    # Mask first installation per group (diff() gives NaN for first row naturally)
    # No need for duplicated() — diff() already returns NaN for first in each group
    valid_intervals = intervals_mins.dropna()
    valid_ids = sorted_df[id_column][valid_intervals.index]

    # Calculate both stats in single groupby
    result = (valid_intervals
              .groupby(valid_ids)
              .agg(
                  app_avg_time_bw_installed_mins_ever='mean',
                  app_median_time_bw_installed_mins_ever='median'
              )
              .reset_index())

    return result


def calculate_installation_intervals_recent(df, id_column, days_before=3):
    """Average and median install gaps for the shared recent-day window."""
    # EDIT 2026-09-30: The old lower-bound-only filter could include apps
    # after the inquiry timestamp. Use the same age mask as category counts.
    mask = df['app_install_age_days'].ge(0) & df['app_install_age_days'].lt(days_before)
    recent_df = (
        df.loc[mask, [id_column, 'first_install_time']]
        .sort_values([id_column, 'first_install_time'])
    )
    intervals_mins = (
        recent_df.groupby(id_column, sort=False)['first_install_time']
        .diff().dt.total_seconds().div(60)
    )
    valid_intervals = intervals_mins.dropna()
    valid_ids = recent_df[id_column][valid_intervals.index]
    return (
        valid_intervals.groupby(valid_ids)
        .agg(**{
            f'app_avg_time_bw_installed_mins_{days_before}d': 'mean',
            f'app_median_time_bw_installed_mins_{days_before}d': 'median',
        })
        .reset_index()
    )
```

## First and last known Play category


```python
def get_edge_category(df, category_column, id_column, install_time_column='first_install_time', get_first=True):
    """
    Find category of first/last installed app for each loan_id efficiently.

    Parameters:
    df (pandas.DataFrame): DataFrame containing:
        - loan_id
        - category_column (column with categories)
        - install_time_column (default: 'first_install_time')
    category_column (str): Name of the column containing categories
    install_time_column (str): Name of the column containing installation times
    get_first (bool): If True, get first installation; if False, get last

    Returns:
    pandas.DataFrame: DataFrame with loan_id and corresponding first/last category
    """
    category_type = 'first' if get_first else 'last'

    # Work only with needed columns, drop nulls in category
    df_work = (df[[id_column, category_column, install_time_column]]
               .dropna(subset=[category_column]))

    # Find edge time per id (min for first, max for last)
    agg_fn = 'min' if get_first else 'max'
    edge_times = df_work.groupby(id_column, sort=False)[install_time_column].agg(agg_fn)

    # Map edge time back via index lookup — no merge/set_index copy
    edge_time_mapped = df_work[id_column].map(edge_times)

    # Filter rows that match the edge time
    edge_df = df_work[df_work[install_time_column] == edge_time_mapped]

    # Handle ties: count category occurrences per id, keep most frequent (alphabetical tie-break)
    result = (edge_df
              .groupby([id_column, category_column], sort=False)
              .size()
              .reset_index(name='_count')
              .sort_values([id_column, '_count', category_column],
                           ascending=[True, False, True])
              .groupby(id_column, sort=False)[category_column]
              .first()
              .reset_index())

    result = result.rename(columns={category_column: f'app_{category_type}_app_cat'})

    return result.fillna(np.nan)
```

## Prefix and merge feature tables


```python
def combine_feature_datasets(base_df, feature_dataset_names, merge_on='loanAccountNumber'):
    """
    Combine base dataset with multiple feature datasets using left join.

    Parameters:
    base_df (pandas.DataFrame): Base DataFrame containing loanAccountNumber and general columns
    feature_dataset_names (list): List of feature dataset variable names
    merge_on (str): Column name to merge on, default 'loanAccountNumber'

    Returns:
    pandas.DataFrame: Combined DataFrame with all features
    """
    # Start with base dataset
    result = base_df.copy()

    # Merge each feature dataset
    for dataset_name in feature_dataset_names:
        # Get the dataset from the variable name
        feature_df = globals()[dataset_name]

        if feature_df.columns.duplicated().any():
            raise ValueError(f'Duplicate columns in {dataset_name}.')
        overlap = (set(result.columns) & set(feature_df.columns)) - {merge_on}
        if overlap:
            raise ValueError(f'Overlapping feature names in {dataset_name}: {sorted(overlap)}')

        # Perform left join
        result = result.merge(
            feature_df,
            on=merge_on,
            how='left',
            validate='1:1',  # Ensure one-to-one merge
            suffixes=('', f'_{dataset_name}')  # Handle duplicate column names
        )

    return result
```

```python
# Preserve the original BigQuery core types and explicit string fields.
core_schema_types = {
    'customerId': 'INT64',
    'digitalLoanAccountId': 'STRING',
    'credolabRefNumber': 'STRING',
    'credo_inquiry_date': 'TIMESTAMP',
    'ln_loan_appln_time': 'TIMESTAMP',
    'Observation_date': 'TIMESTAMP',
    'days_since_credo_call': 'INT64',
}
string_feature_cols = {
    'app_first_app_cat', 'app_last_app_cat', 'app_list', 'package_install_times',
}
numeric_feature_cols = [col for col in feature_cols if col not in string_feature_cols]

# Keep the established integer types of the original Bscore count fields.
legacy_integer_features = ['app_cnt_payday_ever', 'app_cnt_professional_ever', 'app_cnt_religious_ever', 'app_cnt_chinese_non_ps_ever', 'app_cnt_chromium_non_ps_ever', 'app_cnt_chinese_ever', 'app_cnt_lite_ever', 'app_cnt_lite_non_ps_ever', 'app_cnt_paid_ever', 'app_cnt_absence_tag_ever', 'app_cnt_gaming_ever', 'app_cnt_competitors_ever', 'app_cnt_competitors_digital_ever', 'app_cnt_competitors_sil_ever', 'app_cnt_vpn_ever', 'app_cnt_gambling_ever', 'app_cnt_driver_ever', 'app_cnt_nonsystem_ever']
integer_feature_cols = [col for col in numeric_feature_cols if col in legacy_integer_features]
float_feature_cols = [col for col in numeric_feature_cols if col not in legacy_integer_features]

# New numerical fields use FLOAT64, including all-missing recencies and
# count fields affected by left joins. Existing core/count types are retained.
schema_features = [
    bigquery.SchemaField(col, core_schema_types.get(
        col, 'STRING' if col in string_feature_cols
        else 'INT64' if col in legacy_integer_features else 'FLOAT64'
    ))
    for col in output_dt.columns
]

# Keep the pickle frame native; adapt a separate frame for BigQuery.
bq_output_dt = output_dt.copy()
for col in ('customerId', 'days_since_credo_call'):
    bq_output_dt[col] = pd.to_numeric(bq_output_dt[col], errors='raise').astype('Int64')
for col in ('credo_inquiry_date', 'ln_loan_appln_time', 'Observation_date'):
    bq_output_dt[col] = pd.to_datetime(bq_output_dt[col], errors='raise')
for col in ('digitalLoanAccountId', 'credolabRefNumber', *sorted(string_feature_cols)):
    # Nullable strings retain NULL for absent app collections/categories.
    # Non-missing collections retain the original Python-style representation.
    bq_output_dt[col] = bq_output_dt[col].astype('string')
bq_output_dt[float_feature_cols] = (
    bq_output_dt[float_feature_cols].apply(pd.to_numeric, errors='raise').astype('float64')
)
for col in integer_feature_cols:
    bq_output_dt[col] = pd.to_numeric(bq_output_dt[col], errors='raise').astype('Int64')
if np.isinf(bq_output_dt[numeric_feature_cols].to_numpy(dtype='float64', na_value=np.nan)).any():
    raise ValueError('Infinite app-feature values found before export.')
if [field.name for field in schema_features] != bq_output_dt.columns.tolist():
    raise ValueError('BigQuery schema does not match the export columns.')
```
