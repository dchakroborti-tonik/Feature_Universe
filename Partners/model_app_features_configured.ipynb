```python
def save_to_gcs(data, filename, bucket_name):
    """
    Save data to Google Cloud Storage bucket.

    :param data: The data to save. Can be a string, bytes, or a file-like object.
    :param filename: The name of the file to save in the bucket.
    :param bucket_name: The name of the GCS bucket. Defaults to 'ABC'.
    :return: The public URL of the saved file.
    """
    # Create a client
    client = storage.Client()

    # Get the bucket
    bucket = client.bucket(bucket_name)

    # Create a blob (file) in the bucket
    blob = bucket.blob(filename)

    # Determine the content type and upload accordingly
    if isinstance(data, str):
        blob.upload_from_string(data)
    elif isinstance(data, bytes):
        blob.upload_from_string(data, content_type='application/octet-stream')
    elif hasattr(data, 'read'):  # File-like object
        blob.upload_from_file(data)
    else:
        raise ValueError("Unsupported data type. Please provide a string, bytes, or file-like object.")


    print(f"File {filename} uploaded to {bucket_name}.")
    return blob.public_url



def load_pickled_dict_from_gcs(filename, bucket_name, project_id):

    client = storage.Client(project=project_id)

    bucket = client.bucket(bucket_name)
    blob = bucket.blob(filename)

    data = blob.download_as_bytes()

    import pickle
    return pickle.loads(data)



def load_csv_from_gcs(filename, bucket_name):

    return pd.read_csv(
        generate_bucket_url(filename, bucket_name)
    )


def generate_bucket_url(filename, bucket_name):

    return f"gs://{bucket_name}/{filename}"


def union_dict_values(input_dict: dict) -> list:
    """
    Create a unique list of all values from a dictionary of lists.

    Args:
        input_dict (dict): A dictionary where values are lists.

    Returns:
        list: A sorted list of unique values from all lists in the input dictionary.
    """
    # Use a set comprehension to flatten and deduplicate values
    unique_values = {item for sublist in input_dict.values() for item in sublist}

    # Convert the set to a sorted list
    return sorted(unique_values)


def add_prefix_to_columns(df, prefix, columns=None):
    """
    Add a prefix to column names in a pandas DataFrame.

    Parameters:
    df (pandas.DataFrame): The input DataFrame
    prefix (str): The prefix to add to the column names
    columns (list, optional): List of column names to add the prefix to.
                              If None, prefix is added to all columns.

    Returns:
    pandas.DataFrame: A new DataFrame with updated column names
    """
    # If columns is None, use all columns
    if columns is None:
        columns = df.columns

    # Create a dictionary mapping old column names to new prefixed names
    rename_dict = {col: f"{prefix}{col}" for col in columns if col in df.columns}

    # Return a new DataFrame with renamed columns
    return df.rename(columns=rename_dict, inplace=True)


def merge_category_dataframes(base_df, cust_cat_dict):
    """
    Merge dataframes from cust_cat_dict into the base dataset as separate columns.

    Parameters:
    base_df (pandas.DataFrame): The base dataset with 'package_name' column
    cust_cat_dict (dict): Dictionary with tag names as keys and dataframes as values

    Returns:
    pandas.DataFrame: A new DataFrame with additional columns from cust_cat_dict
    """
    if 'package_name' not in base_df.columns:
        raise KeyError("Base DataFrame is missing 'package_name' column")

    result_df = base_df.copy()

    for tag, df in cust_cat_dict.items():
        if 'package_name' not in df.columns:
            raise KeyError(f"DataFrame for tag '{tag}' is missing 'package_name' column")

        if len(df.columns) != 1:
            raise ValueError(f"DataFrame for tag '{tag}' should have only one column (package_name)")

        # Create a new column with the tag name, containing 1 for packages present
        df_merged = df.assign(**{tag: 1})

        # Merge while keeping the 'package_name' column intact
        result_df = result_df.merge(df_merged, on='package_name', how='left')

        # Fill NaN values with 0 and convert to int
        result_df[tag] = result_df[tag].fillna(0).astype(int)

    return result_df


def merge_category_dataframes_optimized(base_df, cust_cat_dict):
    """
    Merge dataframes from cust_cat_dict into the base dataset as separate columns,
    handling potential duplicates without creating a copy of the base DataFrame.

    Parameters:
    base_df (pandas.DataFrame): The base dataset with 'package_name' column
    cust_cat_dict (dict): Dictionary with tag names as keys and dataframes as values

    Returns:
    pandas.DataFrame: The original DataFrame with additional columns from cust_cat_dict
    """
    if 'package_name' not in base_df.columns:
        raise KeyError("Base DataFrame is missing 'package_name' column")

    for tag, df in cust_cat_dict.items():
        if 'package_name' not in df.columns:
            raise KeyError(f"DataFrame for tag '{tag}' is missing 'package_name' column")

        if len(df.columns) != 1:
            raise ValueError(f"DataFrame for tag '{tag}' should have only one column (package_name)")

        # Create a new Series with 1 for packages present, handling duplicates
        category_series = pd.Series(1, index=df['package_name']).groupby(level=0).first()

        # Use map with a default of 0, and convert to int8
        base_df[tag] = base_df['package_name'].map(category_series).fillna(0).astype('int8')

    return base_df
```

## Metadata functions

```python
def add_metadata_columns(df, metadata_dict, metadata_keys, col_prefix):
    """
    Add metadata columns to a DataFrame based on package names, modifying the DataFrame in place.

    Args:
        df (pd.DataFrame): DataFrame with a 'package_name' column.
        metadata_dict (dict): Dictionary of package metadata.
        metadata_keys (list): List of metadata keys to add as columns.
        col_prefix (str): Prefix for new columns

    Warns:
        UserWarning: If any metadata keys are not found.
    """
    # Function to get metadata for a package, returning None if not found
    def get_metadata(package, key):
        return metadata_dict.get(package, {}).get(key)

    # Set to keep track of keys not found in any package
    missing_keys = set(metadata_keys)

    # Calculate total iterations for the progress bar
    total_iterations = len(metadata_keys) * len(df)

    # Create a single progress bar for the entire process
    with tqdm(total=total_iterations, desc="Processing metadata") as pbar:
        # Add new columns for each metadata key
        for key in metadata_keys:
            key_found = False
            new_column = []
            for package in df['package_name']:
                value = get_metadata(package, key)
                new_column.append(value)
                if value is not None:
                    key_found = True
                pbar.update(1)  # Update progress bar for each package

            df[f'{col_prefix}{key}'] = new_column
            if key_found:
                missing_keys.discard(key)

    # Warn about keys not found in any package
    if missing_keys:
        warnings.warn(f"The following keys were not found in any package's metadata: {', '.join(missing_keys)}")


def extract_first_category_id(df, column_name):
    """
    Extract the ID of the first category from a column containing a list of category dictionaries.

    Args:
        df (pd.DataFrame): Input DataFrame.
        column_name (str): Name of the column containing category information.

    Returns:
        pd.Series: Series with the first category ID or Unknown (in case of category ID = None) or None values.
    """
    error_count = 0

    def get_first_id(value):
        nonlocal error_count
        try:
            if value:
                result = value[0]['id'].lower()
                if result:
                    return result
                else:
                    return "Unknown"
            else:
                return None
        except:
            # Increment error count and return NA if there's any error in parsing or accessing the data
            error_count += 1
            return None

    result = df[column_name].apply(get_first_id)

    print(f"Total errors encountered during parsing or accessing data: {error_count}")

    return result

def extract_first_category_name(df, column_name):
    """
    Extract the name of the first category from a column containing a list of category dictionaries.

    Args:
        df (pd.DataFrame): Input DataFrame.
        column_name (str): Name of the column containing category information.

    Returns:
        pd.Series: Series with the first category name or Unknown (in case of category name = None) or None values.
    """
    error_count = 0

    def get_first_name(value):
        nonlocal error_count
        try:
            if value:
                result = value[0]['name'].lower()
                if result:
                    return result
                else:
                    return "Unknown"
            else:
                return None
        except:
            # Increment error count and return NA if there's any error in parsing or accessing the data
            error_count += 1
            return None

    result = df[column_name].apply(get_first_name)

    print(f"Total errors encountered during parsing or accessing data: {error_count}")

    return result
```

## ODR functions

```python
def calc_odr_table(df, column):
    """
    Calculate ODR (Observed Default Rate) table for a given column.

    Args:
    df (pd.DataFrame): Input DataFrame containing loan data.
    column (str): Column name to group by for ODR calculation.

    Returns:
    pd.DataFrame: ODR table with package counts, loan counts, default rates, and loan share.
    """
    total_loans = df['loanAccountNumber'].nunique()
    return (
        df.groupby([column])
        .agg({'package_name': 'nunique', 'loanAccountNumber': 'nunique', 'deffspd30': 'mean'})
        .assign(share=lambda x: x['loanAccountNumber'] / total_loans)
        .rename(columns={'package_name':'cnt_unique_packages', 'loanAccountNumber': 'cnt_loans'})
        .sort_values(['cnt_unique_packages'], ascending=False)
    )
```

```python
# Only for Research, remove for final version
def calc_odr_table_multi_category(df, category_columns):
    """
    Calculate ODR (Observed Default Rate) table for multiple category columns.

    Args:
    df (pd.DataFrame): Input DataFrame containing loan data and category columns.
    category_columns (list): List of column names representing different categories.

    Returns:
    pd.DataFrame: ODR table with package counts, loan counts, default rates, and loan share for each category.
    """
    total_loans = df['loanAccountNumber'].nunique()

    # Create a list to store results for each category
    results = []

    for column in category_columns:
        # For each category column, create a temporary DataFrame with loans that belong to this category
        temp_df = df[df[column] == 1]

        category_result = pd.DataFrame({
            'category': [column],
            'cnt_unique_packages': [temp_df['package_name'].nunique()],
            'cnt_loans': [temp_df['loanAccountNumber'].nunique()],
            'deffspd30': [temp_df['deffspd30'].mean()],
        })

        results.append(category_result)

    # Combine results from all categories
    result_df = pd.concat(results, ignore_index=True)

    # Calculate the share
    result_df['share'] = result_df['cnt_loans'] / total_loans

    # Sort by cnt_unique_packages in descending order
    result_df = result_df.sort_values('cnt_unique_packages', ascending=False)
    result_df = result_df.set_index('category')

    return result_df
```

```python
def check_app_status_vectorized(flags):
    """
    Vectorized function to determine if applications are system (1) or non-system (0)
    based on their flag values. Handles NA values by returning np.nan.

    Args:
        flags (pd.Series or np.array or list or int): Application flag values

    Returns:
        pd.Series or int: 1 (system) for odd numbers, 0 (non-system) for even numbers,
                         np.nan for NA/null values
    """
    if isinstance(flags, (pd.Series, np.ndarray, list)):
        # Convert to pandas Series if not already
        flags = pd.Series(flags)

        # Initialize result series with NaN values
        result = pd.Series([np.nan] * len(flags), index=flags.index)

        # Update non-null values
        mask = flags.notna()
        result[mask] = np.where(flags[mask].astype(float) % 2 == 1, 1, 0)

        return result
    else:
        # Handle single value
        if pd.isna(flags):
            return np.nan
        return 1 if int(flags) % 2 == 1 else 0
```

```python
def create_mapping_dict(columns, pattern='app_cnt_{name}_ever'):
    """
    Replace '*_cat_' or 'cat_' prefix with pattern in column names.

    Parameters:
    columns (list): List of column names
    pattern (str): Pattern to use for new names, must contain {name} placeholder
        default: 'app_cnt_{name}_ever'

    Returns:
    dict: Dictionary mapping old column names to new ones
    """
    rename_dict = {}

    for col in columns:
        if '_cat_' in col:
            new_name = col.split('_cat_')[1]
            new_name = pattern.format(name=new_name)
            rename_dict[col] = new_name
        elif col.startswith('cat_'):
            new_name = col.replace('cat_', '', 1)  # just remove 'cat_'
            new_name = pattern.format(name=new_name)
            rename_dict[col] = new_name
        else:
            rename_dict[col] = col

    return rename_dict
```

```python
def get_first_values_by_loan(df, category_columns, id_column):
    """
    Get first values from multiple columns for each loan.

    Parameters:
    df (pandas.DataFrame): DataFrame containing:
        - id_column
        - columns (columns to get first values from)
    columns (list): List of column names to get first values

    Returns:
    pandas.DataFrame: DataFrame with first values from specified columns for each loan
    """
    return df.groupby(id_column)[category_columns].first().reset_index()
```

## General categories (count)

```python
def aggregate_categories_single_column(df, col_with_categories, id_column):
    """
    Aggregate package counts by category (single column) at the loan level.

    Parameters:
    df (pandas.DataFrame): DataFrame containing:
        - id_column
        - package_name
        - col_with_categories

    Returns:
    pandas.DataFrame: Aggregated DataFrame with:
        - id_column as index
        - Columns for each general category containing package counts
    """
    # First, get unique combinations of loan and category with their counts
    category_counts = (df.groupby([id_column, col_with_categories])
                      .package_name
                      .count()
                      .reset_index())

    # Pivot the results to get categories as columns
    result = category_counts.pivot(
        index=id_column,
        columns=col_with_categories,
        values='package_name'
    ).fillna(0)

    # Convert counts to integers
    result = result.astype(int)

    # Reset index to make id_column a column
    result = result.reset_index()

    # Rename columns to avoid potential naming conflicts
    category_columns = {col: f'app_cnt_{col}_ever' for col in result.columns
                       if col != id_column}
    result = result.rename(columns=category_columns)

    return result.reset_index(drop=True)
```

```python
def aggregate_categories_different_columns_in_period(df, category_columns, id_column, days_before=7):
    """
    Aggregate package counts by customer categories at the loan level,
    counting only packages installed within specified days before loan application.

    Parameters:
    df (pandas.DataFrame): DataFrame containing:
        - id_column
        - credo_inquiry_date
        - category columns (binary 0/1 values)
    category_columns (list): List of category column names to aggregate
    days_before (int): Number of days before application to consider

    Returns:
    pandas.DataFrame: Aggregated DataFrame with counts for each category
    """
    # Calculate time window for each loan
    window_start = df['credo_inquiry_date'] - pd.Timedelta(days=days_before)

    # Filter rows within the time window
    mask = (df['first_install_time'] >= window_start) & \
           (df['first_install_time'] <= df['credo_inquiry_date'])

    return df[mask].groupby(id_column)[category_columns].sum().reset_index()
```

### Max age of app from specific category

```python
def calculate_time_since_first_app(df, category_column, id_column,
                                   result_column_name='days_since_first_app'):
    """
    Calculate time difference in days between first specific app installation
    and loan application submission for each loan.

    Parameters:
    df (pandas.DataFrame): DataFrame containing:
        - id_column
        - credo_inquiry_date
        - first_install_time
        - category_column (1/0)
        - result_column_name - (optional) column name in return; default 'days_since_first_app'

    Returns:
    pandas.DataFrame: DataFrame with id_column and result_column_name
    """
    # Get minimum install time per id for rows where category == 1
    first_installs = (
        df.loc[df[category_column] == 1, [id_column, 'first_install_time']]
        .groupby(id_column, sort=False)['first_install_time']
        .min()
    )

    # Get unique application dates per id
    app_times = (
        df[[id_column, 'credo_inquiry_date']]
        .drop_duplicates(subset=id_column)
        .set_index(id_column)['credo_inquiry_date']
    )

    # Align app_times to first_installs index (preserves left-join behavior)
    app_times_aligned = app_times.reindex(first_installs.index)

    # Calculate difference in days
    result = ((app_times_aligned - first_installs)
              .dt.total_seconds()
              .div(86400)
              .round(1))

    return result.rename(result_column_name).reset_index()
```

### Average/median installation time difference

```python
def calculate_installation_intervals_alltime(df, id_column):
    """
    Calculate average and median time between app installations for all time,
    excluding the first (oldest) installation time.

    Parameters:
    df (pandas.DataFrame): DataFrame containing:
        - id_column
        - first_install_time

    Returns:
    pandas.DataFrame: DataFrame with intervals calculations
    """
    # Sort only necessary columns to reduce memory footprint
    sorted_df = (df[[id_column, 'first_install_time']]
                 .sort_values([id_column, 'first_install_time']))

    # Calculate time differences and convert to minutes in one step
    intervals_mins = (sorted_df
                      .groupby(id_column, sort=False)['first_install_time']
                      .diff()
                      .dt.total_seconds()
                      .div(60))

    # Mask first installation per group (diff() gives NaN for first row naturally)
    # No need for duplicated() — diff() already returns NaN for first in each group
    valid_intervals = intervals_mins.dropna()
    valid_ids = sorted_df[id_column][valid_intervals.index]

    # Calculate both stats in single groupby
    result = (valid_intervals
              .groupby(valid_ids)
              .agg(
                  app_avg_time_bw_installed_mins_ever='mean',
                  app_median_time_bw_installed_mins_ever='median'
              )
              .reset_index())

    return result


def calculate_installation_intervals_recent(df, id_column, days_before=3):
    """
    Calculate average and median time between recent app installations.

    Parameters:
    df (pandas.DataFrame): DataFrame containing:
        - id_column
        - first_install_time
        - credo_inquiry_date
    days_before (int): Number of days before application to consider

    Returns:
    pandas.DataFrame: DataFrame with intervals calculations
    """
    # Compute period_start per id without merging back to full df
    period_start = (df.groupby(id_column, sort=False)['credo_inquiry_date']
                    .first()
                    .sub(pd.Timedelta(days=days_before)))

    # Map period_start via index lookup instead of merge (no full copy)
    install_times = df[[id_column, 'first_install_time']].copy()
    install_times['period_start'] = install_times[id_column].map(period_start)

    # Filter recent installations and drop helper column
    recent_df = (install_times[install_times['first_install_time'] >= install_times['period_start']]
                 .drop(columns='period_start')
                 .sort_values([id_column, 'first_install_time']))

    # Calculate intervals in minutes — NaN naturally marks first per group
    intervals_mins = (recent_df
                      .groupby(id_column, sort=False)['first_install_time']
                      .diff()
                      .dt.total_seconds()
                      .div(60))

    valid_intervals = intervals_mins.dropna()
    valid_ids = recent_df[id_column][valid_intervals.index]

    # Single groupby for both stats
    result = (valid_intervals
              .groupby(valid_ids)
              .agg(**{
                  f'app_avg_time_bw_installed_mins_{days_before}d': 'mean',
                  f'app_median_time_bw_installed_mins_{days_before}d': 'median'
              })
              .reset_index())

    return result
```

### Category of first/last ever non-system PS app

```python
def get_edge_category(df, category_column, id_column, install_time_column='first_install_time', get_first=True):
    """
    Find category of first/last installed app for each loan_id efficiently.

    Parameters:
    df (pandas.DataFrame): DataFrame containing:
        - loan_id
        - category_column (column with categories)
        - install_time_column (default: 'first_install_time')
    category_column (str): Name of the column containing categories
    install_time_column (str): Name of the column containing installation times
    get_first (bool): If True, get first installation; if False, get last

    Returns:
    pandas.DataFrame: DataFrame with loan_id and corresponding first/last category
    """
    category_type = 'first' if get_first else 'last'

    # Work only with needed columns, drop nulls in category
    df_work = (df[[id_column, category_column, install_time_column]]
               .dropna(subset=[category_column]))

    # Find edge time per id (min for first, max for last)
    agg_fn = 'min' if get_first else 'max'
    edge_times = df_work.groupby(id_column, sort=False)[install_time_column].agg(agg_fn)

    # Map edge time back via index lookup — no merge/set_index copy
    edge_time_mapped = df_work[id_column].map(edge_times)

    # Filter rows that match the edge time
    edge_df = df_work[df_work[install_time_column] == edge_time_mapped]

    # Handle ties: count category occurrences per id, keep most frequent (alphabetical tie-break)
    result = (edge_df
              .groupby([id_column, category_column], sort=False)
              .size()
              .reset_index(name='_count')
              .sort_values([id_column, '_count', category_column],
                           ascending=[True, False, True])
              .groupby(id_column, sort=False)[category_column]
              .first()
              .reset_index())

    result = result.rename(columns={category_column: f'app_{category_type}_app_cat'})

    return result.fillna(np.nan)
```

## Features merging

```python
def combine_feature_datasets(base_df, feature_dataset_names, merge_on='loanAccountNumber'):
    """
    Combine base dataset with multiple feature datasets using left join.

    Parameters:
    base_df (pandas.DataFrame): Base DataFrame containing loanAccountNumber and general columns
    feature_dataset_names (list): List of feature dataset variable names
    merge_on (str): Column name to merge on, default 'loanAccountNumber'

    Returns:
    pandas.DataFrame: Combined DataFrame with all features
    """
    # Start with base dataset
    result = base_df.copy()

    # Merge each feature dataset
    for dataset_name in feature_dataset_names:
        # Get the dataset from the variable name
        feature_df = globals()[dataset_name]

        # Perform left join
        result = result.merge(
            feature_df,
            on=merge_on,
            how='left',
            validate='1:1',  # Ensure one-to-one merge
            suffixes=('', f'_{dataset_name}')  # Handle duplicate column names
        )

    return result
```
