# B-Score MoEngage Event Features
**Version**: v20260818 | **Model**: B-Score | **Date**: 2026-08-18

> **v20260818 changes vs v20260428**
> - `digitalLoanAccountId` is **not** read, not used for grouping, and not written to output.
> - Grain is `(customer_id, observation_datetime)` throughout — no loan-level step, no dedup step.
> - Source table is the pre-built snapshot `risk_mart.bscore_moenagage_events_all_20230101_v6_snapshot_20260817`; the 3 BQ creation queries from v20260428 are removed.
> - Feature package import changed to `B_Score_Events_FE_codebase_v20260818`.
> - `event_time_manila` and `observation_datetime` are both tz-naive throughout (no `tz_localize` calls).

**Pipeline:**
1. Load pre-built events snapshot from BigQuery (single read, 5 columns, no `digitalLoanAccountId`)
2. Preprocess: UTC → Manila tz-naive, `observation_datetime = Observation_date + 23:59:00`
3. Build one metadata row per `(customer_id, observation_datetime)`
4. Pre-group events by `(customer_id, observation_datetime)` for O(1) lookup
5. Compute ~300 features per customer-observation pair via 11 feature modules (**parallel**)
6. Write to BigQuery + local CSV backup

**Performance**: Pre-grouped O(1) pair lookup + `ThreadPoolExecutor` parallel processing.

## 6  Preprocess Timestamps
- `event_time` (UTC) → `event_time_manila` (Asia/Manila, **tz-naive**)
- `Observation_date` (Manila date) → `observation_datetime` = date + 23:59:00 (**tz-naive**)

Both columns are tz-naive after this step — matching the contract expected by `B_Score_Events_FE_codebase_v20260818`.

```python
def preprocess(df: pd.DataFrame) -> pd.DataFrame:
    """
    1. event_time (UTC) -> event_time_manila (Asia/Manila, tz-naive)
    2. Observation_date (Manila date) -> observation_datetime = date + 23:59:00 (tz-naive)

    No digitalLoanAccountId column in v20260818.
    """
    df = df.copy()

    df['event_time'] = pd.to_datetime(df['event_time'], utc=True, errors='coerce')
    df['event_time_manila'] = (
        df['event_time']
        .dt.tz_convert('Asia/Manila')
        .dt.tz_localize(None)
    )
    df.drop(columns=['event_time'], inplace=True)

    df['Observation_date'] = pd.to_datetime(df['Observation_date'], errors='coerce')
    df['observation_datetime'] = (
        df['Observation_date'] + pd.Timedelta(hours=23, minutes=59, seconds=0)
    )
    df.drop(columns=['Observation_date'], inplace=True)

    df['customer_id'] = df['customer_id'].astype(str)
    return df


events_df = preprocess(raw_df)
del raw_df   # free raw copy

print(f'Preprocessed shape   : {events_df.shape}')
print(f'Columns              : {events_df.columns.tolist()}')
print(f'event_time_manila    : {events_df["event_time_manila"].dtype}')
print(f'observation_datetime : {events_df["observation_datetime"].dtype}')
events_df[['event_time_manila', 'observation_datetime']].head(3)
```

### 7b  Build Customer-Observation Metadata & Pre-Group Events
v20260818 grain: **`(customer_id, observation_datetime)`** — no `digitalLoanAccountId`.  
One metadata row per unique pair; events pre-grouped by the same key for O(1) lookup.

```python
def build_customer_obs_meta(df: pd.DataFrame) -> pd.DataFrame:
    """One row per unique (customer_id, observation_datetime).

    moengage_user_id is taken from the first row in each pair
    (consistent with main.py build_customer_obs_meta).
    """
    return (
        df.groupby(['customer_id', 'observation_datetime'], sort=False)
        .agg(
            moengage_user_id=('moengage_user_id', 'first'),
        )
        .reset_index()
    )


customer_obs_meta = build_customer_obs_meta(events_df)

# Pre-group by (customer_id, observation_datetime) for O(1) per-pair event lookup
grouped = events_df.groupby(['customer_id', 'observation_datetime'], sort=False)

print(f'Total customer-observation pairs : {len(customer_obs_meta):,}')
print(f'Unique customers                 : {events_df["customer_id"].nunique():,}')
print(f'Unique observation datetimes     : {events_df["observation_datetime"].nunique():,}')
customer_obs_meta.head(3)
```

### 7c  Parallel Feature Generation
Uses `ThreadPoolExecutor` for concurrent pair processing.  
Pre-grouped DataFrame ensures O(1) event lookup per `(customer_id, observation_datetime)`.

No `digitalLoanAccountId` in the row dict — output is already at the correct grain; **no dedup step needed**.

```python
# ── Worker function ────────────────────────────────────────────────────────────
def process_single_pair(cust_id, obs_dt, moe_uid, pair_events_df):
    """Apply 7-day window filter, compute all features, return result dict."""
    row_base = {
        'customer_id'         : cust_id,
        'moengage_user_id'    : moe_uid,
        'observation_datetime': obs_dt,
    }
    try:
        window_start = obs_dt - pd.Timedelta(hours=168)
        window_df = pair_events_df[
            (pair_events_df['event_time_manila'] >= window_start) &
            (pair_events_df['event_time_manila'] <= obs_dt)
        ].copy()
        window_df.sort_values('event_time_manila', inplace=True)
        window_df['observation_datetime'] = obs_dt

        feats = create_all_features(window_df) if len(window_df) > 0 else {}
        row_base.update(feats)
    except Exception as exc:
        print(f'  ERROR  cust={cust_id}  obs={obs_dt}: {exc}')
    return row_base


# ── Parallel execution ─────────────────────────────────────────────────────────
N_WORKERS = min(8, (os.cpu_count() or 4))
print(f'Running with {N_WORKERS} worker threads across {len(customer_obs_meta):,} pairs ...')

all_results = []
t0 = time.time()

with ThreadPoolExecutor(max_workers=N_WORKERS) as executor:
    futures = {}
    for row in customer_obs_meta.itertuples(index=False):
        cust_id = row.customer_id
        obs_dt  = row.observation_datetime
        moe_uid = row.moengage_user_id
        try:
            pair_events = grouped.get_group((cust_id, obs_dt))
        except KeyError:
            pair_events = pd.DataFrame(columns=events_df.columns)
        futures[executor.submit(
            process_single_pair,
            cust_id, obs_dt, moe_uid, pair_events
        )] = (cust_id, obs_dt)

    for future in tqdm(
        as_completed(futures),
        total=len(futures),
        desc='Processing customer-observation pairs',
    ):
        result = future.result()
        if result:
            all_results.append(result)

elapsed = time.time() - t0
print(f'Processed {len(all_results):,} pairs in {elapsed:.1f}s '
      f'({len(all_results) / elapsed:.1f} pairs/sec)')
```

### 7d  Assemble Results DataFrame
Output grain is already `(customer_id, observation_datetime)` — **no dedup step required** in v20260818.  
Column order: `customer_id | moengage_user_id | observation_datetime | …features…`

```python
results_df = pd.DataFrame(all_results)

# Ensure identifier columns are first
id_cols   = ['customer_id', 'moengage_user_id', 'observation_datetime']
feat_cols = [c for c in results_df.columns if c not in id_cols]
results_df = results_df[id_cols + feat_cols]

# Sanity: one row per (customer_id, observation_datetime) — no duplicates expected
dupe_count = results_df.groupby(['customer_id', 'observation_datetime']).size().max()
assert dupe_count == 1, (
    f'Unexpected duplicate (customer_id, observation_datetime) rows! '
    f'max group size = {dupe_count}'
)

print(f'Shape (customer-observation pairs): {results_df.shape}')
print(f'Feature columns                   : {len(feat_cols)}')
results_df[id_cols].head(3)
```
